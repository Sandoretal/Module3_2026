```python
# NOTE: don't worry about this cell. Run it once: it creates the two data files for this task.
import os
import numpy as np
import pandas as pd

os.makedirs("end_of_day_data", exist_ok=True)

# 1. Health records (same as in the pandas tutorial)
pd.DataFrame({
    "patient_id": ["P01", "P02", "P03", "P04", "P05", "P01", "P02", "P03", "P05", "P06", "P01", "P02", "P05", "P04"],
    "sex": ["Male", "Female", "Female", "Male", "Male", "Male", "Female", "Female", "Male", "Female", "Male", "Female", "Male", "Male"],
    "date_of_birth": ["1950-05-14", "1965-08-22", "1970-11-30", "1980-01-15", "1958-03-02", "1950-05-14", "1965-08-22",
                      "1970-11-30", "1958-03-02", "1955-12-01", "1950-05-14", "1965-08-22", "1958-03-02", "1980-01-15"],
    "visit_date": ["2023-01-10", "2023-01-15", "2023-02-01", "2023-03-14", "2023-04-21", "2023-07-12", "2023-09-03",
                   "2023-10-06", "2023-11-30", "2024-01-09", "2024-02-08", "2024-03-20", "2024-05-17", "2024-06-04"],
    "clinic_id": [101, 102, 101, 103, 102, 101, 102, 101, 102, 103, 101, 102, 102, 103],
    "updrs_iii": [22, 30, 40, 12, 35, 25, None, 42, 38, 27, 28, 31, 41, None],
    "ledd_mg": [300, 150, 800, 0, 450, 400, 300, 850, 600, 400, 500, 300, 750, 0],
    "medication": ["Co-careldopa", "Rotigotine", "Co-beneldopa", None, "Pramipexole", "co-careldopa ", "rotigotine",
                   "co-beneldopa", "Co-careldopa + Entacapone", "Ropinirole", "CO-CARELDOPA", " Rotigotine",
                   "co-careldopa + entacapone ", None],
    "bmi": ["22.5", "25.0", "30.0", "28.0", "26.4", "22.7", "not recorded", "31.2", "26.1", "21.9", "23.1", "24.0",
            "not recorded", "27.6"],
}).to_csv("data/ehr_parkinsons.csv", index=False)

# 2. Wrist accelerometer: 20 s at rest, 100 Hz, one recording per patient
rng = np.random.default_rng(seed=7)
fs = 100
t = np.arange(20 * fs) / fs
tremor_settings = {"P01": (5.0, 0.035), "P02": (4.4, 0.025), "P03": (5.4, 0.090),
                   "P04": (0.0, 0.000), "P05": (4.8, 0.070), "P06": (5.8, 0.030)}   # (frequency in Hz, amplitude in g)
recordings = []
for pid, (freq, amp) in tremor_settings.items():
    phase = rng.uniform(0, 2 * np.pi)
    tremor = amp * (1 + 0.3 * np.sin(2 * np.pi * 0.1 * t + phase)) * np.sin(2 * np.pi * freq * t + phase)
    drift = 0.02 * np.sin(2 * np.pi * 0.15 * t + phase)          # slow changes in wrist posture
    rec = pd.DataFrame({
        "patient_id": pid,
        "t_s": np.round(t, 2),
        "Acc_X": 0.10 + 0.8 * tremor + drift + rng.normal(0, 0.005, t.size),
        "Acc_Y": 0.15 + 0.4 * tremor + rng.normal(0, 0.005, t.size),
        "Acc_Z": 0.98 + 0.6 * tremor + rng.normal(0, 0.005, t.size),
        "fs_hz": fs,
    })
    if pid == "P03":   # a 0.6-second dropout
        rec.loc[800:859, ["Acc_X", "Acc_Y", "Acc_Z"]] = np.nan
    if pid == "P05":   # 20 scattered missing samples
        rec.loc[rng.choice(np.arange(10, 1990), 20, replace=False), ["Acc_X", "Acc_Y", "Acc_Z"]] = np.nan
    recordings.append(rec)

pd.concat(recordings, ignore_index=True).round({"Acc_X": 4, "Acc_Y": 4, "Acc_Z": 4}).to_csv(
    "data/wearable_rest_recordings.csv", index=False)

print("Files created in the folder 'end_of_day_data'!")
```

```python
# ONE POSSIBLE SOLUTION (other correct solutions exist, e.g. forward fill instead of interpolation)

# ======================= PART A: the health records =======================
# 1. Load and convert the dates
ehr = pd.read_csv("data/ehr_parkinsons.csv")
ehr["visit_date"] = pd.to_datetime(ehr["visit_date"])

# 2. Each patient's latest visit date on every row (.transform), then keep the rows that match it
ehr["last_visit"] = ehr.groupby("patient_id")["visit_date"].transform("max")
ehr_last = ehr[ehr["visit_date"] == ehr["last_visit"]]

# 3. Keep only the columns we need
ehr_last = ehr_last[["patient_id", "visit_date", "updrs_iii", "ledd_mg", "medication"]]
display(ehr_last)

# ======================= PART B: from signals to features =======================
# 1. Load and inspect
rec = pd.read_csv("data/wearable_rest_recordings.csv")
print("Shape (rows, columns):", rec.shape)
print("Number of patients:", rec["patient_id"].nunique())
print("Seconds per recording:")
display(rec.groupby("patient_id").size() / 100)

# 2. Missing values: how many, and whose?
print("Missing values per column:")
display(rec.isna().sum())
display(rec[rec["Acc_X"].isna()]["patient_id"].value_counts())

# 3. Fill in the gaps (linear interpolation; rec[acc_columns].ffill() would also work)
acc_columns = ["Acc_X", "Acc_Y", "Acc_Z"]
rec[acc_columns] = rec[acc_columns].interpolate()
print("Missing values after filling:")
display(rec[acc_columns].isna().sum())

# 4. Signal vector magnitude
rec["vector_mag"] = np.sqrt(rec["Acc_X"]**2 + rec["Acc_Y"]**2 + rec["Acc_Z"]**2)

# 5. One row per patient with two features (.agg with a built-in name AND with our own function)
def iqr(values):
    return values.quantile(0.75) - values.quantile(0.25)

features = rec.groupby("patient_id").agg(
    tremor_amplitude=("vector_mag", "std"),
    dominant_freq_hz=("vector_mag", iqr)
).reset_index()
display(features)

# ======================= PART C: combine and interpret =======================
# 1. Merge the features with the most recent visit of each patient
combined = features.merge(ehr_last, on="patient_id", how="left")
display(combined)

# 2. Scatter plot (P04 has no UPDRS III at this visit, so we keep only the patients that have one)
plot_data = combined.dropna(subset=["updrs_iii"])

plt.figure(figsize=(7, 4))
plt.scatter(plot_data["tremor_amplitude"], plot_data["updrs_iii"])


# 4: is the dominant frequency in the typical range of a Parkinsonian rest tremor?
combined["pd_tremor_range"] = combined["dominant_freq_hz"].apply(lambda freq: 4 <= freq <= 6)
display(combined[["patient_id", "tremor_amplitude", "dominant_freq_hz", "pd_tremor_range"]])


# ======================= Answers to 3 =======================
# 1. P03 has the strongest tremor (about 0.047 g), followed by P05. At that visit P03 was on co-beneldopa
#    (a levodopa-based drug) with an LEDD of 850 mg/day, the highest dose in the table.
# 2. Roughly yes: P03 and P05 have both the strongest tremor and the highest UPDRS III scores (42 and 41).
#    But five patients are far too few to draw any conclusion, and the relationship is not perfect: P02 has a
#    higher score (31) than P01 (28) and P06 (27), even though P02's tremor is weaker. UPDRS III also scores rigidity,
#    slowness of movement, gait and posture, so a patient can have a high score with little tremor.
# 3. P04: their UPDRS III score is missing (NaN) at their most recent visit, so there is no y value to plot.
#    (P04 is also the only untreated patient: ledd_mg = 0.)
```
